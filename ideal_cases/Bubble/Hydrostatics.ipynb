```python
import numpy as np

cp = 1004.
g  = 9.81
t00 = 300.
gamma = 1.4
gamma2 =  gamma / (gamma - 1.0) 


h2 = cp*t00/g

print(h2)
```

```python
def scale(z, h):
    return (1.0 - z/h)
def tprofile(t, z, h):
    return scale(z,h)*t
def pprofile(p, z, h):
    return p*(scale(z,h)**gamma2)
```

```python
print(tprofile(303., 6e3, h2))
print(pprofile(1000., 6e3, h2))
```
